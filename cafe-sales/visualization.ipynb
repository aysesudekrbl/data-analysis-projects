*Data Quality Note:* Location is unknown for about 40% of transactions, and Payment Method is unknown for about 32%. This means any pattern based on these two columns should be read with caution.

```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
df = pd.read_csv("cafe_sales_cleaned.csv")
```

```python
df. head()
```

```python
sns.set_theme(style = "darkgrid")
```

```python
sns.set_palette(sns.color_palette("bright6"))
```

```python
sns.displot(df, x = "Item", kind = "hist", hue = "Payment Method", col = "Location")
```

_Across all three locations, the 'Unknown' item category is clearly the least common, while the other items (Tea, Salad, Cookie, Juice, Smoothie, Coffee, Cake, Sandwich) have similar counts to each other."_

```python
(df["Location"] == "Unknown").mean()*100
```

```python
(df["Payment Method"] == "Unknown").mean()*100
```

```python
sns.countplot(df, x= "Item", hue = "Payment Method")
```

_For every item, 'Unknown' is the most common payment method, more frequent than Credit Card, Cash, or Digital Wallet individually. This suggests the missing payment data isn't random, but may point to a systematic gap in how transactions were recorded."_
