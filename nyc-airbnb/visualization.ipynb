```python
import seaborn as sns
```

```python
import pandas as pd
```

```python
df = pd.read_csv("AB_NYC_2019_Cleaned.csv")
```

```python
df.head()
```

```python
sns.set_theme(style = "darkgrid")
```

```python
sns.set_palette(sns.color_palette("bright6"))
```

```python
sns.barplot(df,x = "neighbourhood_group", y = "price", hue="room_type")
```

_In every borough, the price order is the same: Entire home/apt is most expensive, then Private room, then Shared room, confirming that room type and location affect price independently. However, the error bars for Staten Island and Bronx are much longer, meaning their averages are less reliable due to having far fewer listings._

```python
sns.relplot(df, x = "number_of_reviews", y= "price", col = "room_type", kind = "scatter")
import matplotlib.pyplot as plt
plt.ylim(0, 1000)
```

_After zooming into the 0-1000 price range, the pattern is clearer: reviews are concentrated at low review counts across all price levels, and listings with many reviews (200+) are almost always priced under 200. Expensive listings tend to have very few reviews. This explains the near-zero correlation found earlier and the relationship isn't linear._

```python
sns.countplot(df, x="neighbourhood_group")
```

_Manhattan and Brooklyn have by far the most listings, while Staten Island and Bronx have very few. This explains why the price averages for Staten Island and Bronx in the earlier chart had such long error bars — they're based on a much smaller sample._
